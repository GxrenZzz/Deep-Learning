## Genearating a nn to learn notation of weights bias and outputs 

```python
from graphviz import Digraph

dot = Digraph("NeuralNetwork")
dot.attr(rankdir="LR")  # Left to Right

# Input Layer
with dot.subgraph(name="cluster_input") as c:
    c.attr(label="Input Layer")
    for i in range(1, 6):
        c.node(f"x{i}", f"x{i}", shape="circle")

# Hidden Layer 1
with dot.subgraph(name="cluster_h1") as c:
    c.attr(label="Hidden Layer 1")
    for i in range(1, 5):
        c.node(f"h1{i}", "", shape="circle")

# Hidden Layer 2
with dot.subgraph(name="cluster_h2") as c:
    c.attr(label="Hidden Layer 2")
    for i in range(1, 4):
        c.node(f"h2{i}", "", shape="circle")

# Hidden Layer 3
with dot.subgraph(name="cluster_h3") as c:
    c.attr(label="Hidden Layer 3")
    for i in range(1, 3):
        c.node(f"h3{i}", "", shape="circle")

# Output Layer
with dot.subgraph(name="cluster_out") as c:
    c.attr(label="Output Layer")
    for i in range(1, 3):
        c.node(f"o{i}", f"y{i}", shape="circle")

# Fully connect layers
for i in range(1, 6):
    for j in range(1, 5):
        dot.edge(f"x{i}", f"h1{j}")

for i in range(1, 5):
    for j in range(1, 4):
        dot.edge(f"h1{i}", f"h2{j}")

for i in range(1, 4):
    for j in range(1, 3):
        dot.edge(f"h2{i}", f"h3{j}")

for i in range(1, 3):
    for j in range(1, 3):
        dot.edge(f"h3{i}", f"o{j}")

# Save and display
dot.render("neural_network", format="png", view=True)
```
