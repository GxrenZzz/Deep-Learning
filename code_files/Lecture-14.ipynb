```python
df = df.drop(columns=["Serial No."])
```

```python
# test-train split 

X = df.iloc[:,0:-1]
y = df.iloc[:,-1]

from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=1)

# min-max scaling -> when u know upper bound and lower bound 

from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler () 

X_train_scale = scaler.fit_transform(X_train)
X_test_scale = scaler.transform(X_test)
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model = Sequential()

model.add(Dense(7,activation='relu',input_dim=7))
model.add(Dense(7,activation='relu'))
model.add(Dense(1,activation='linear'))
```

```python
model.summary()
```

```python
# compiling the model to configiure 

model.compile(loss='mean_squared_error',optimizer='Adam',metrics=['accuracy'])

# train the model 

hisory = model.fit(X_train_scale,y_train,epochs=100,validation_split=0.2)
```

```python
y_pred = model.predict(X_test_scale)
```

```python
from sklearn.metrics import r2_score

r2_score(y_test,y_pred)
```

```python
import matplotlib.pyplot as plt 

plt.plot(hisory.history['loss'])
plt.plot(hisory.history['val_loss'])
```
