```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
# for plotting the region we use mlxtend lib
from mlxtend.plotting import plot_decision_regions
```

```python
df = pd.read_csv("/Users/piyush/Repositories/Deep-Learning/Datasets/placement.csv")
```

```python
print(df.shape)
df.head()
```

```python
# Plotting the data on scatterplot 

sns.scatterplot(
    x=df['cgpa'],
    y=df['resume_score'],
    hue=df['placed']
)
```

```python
# Splitting the data set 
X = df.iloc[:,0:2]
y = df.iloc[:,-1]
```

```python
print(X.shape)
print(y.shape)
```

```python
# Importing scikit_learn lib to compute perceptron
from sklearn.linear_model import Perceptron
p = Perceptron()
```

```python
p.fit(X,y)
```

```python
p.coef_
```

```python
p.intercept_
```

```python
# plotting the reigons 
# clf -> classifier 
plot_decision_regions(X.values,y.values, clf=p , legend = 2)
```
